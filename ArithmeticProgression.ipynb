# Q1. Can Make Arithmetic Progression From Sequence

A sequence of numbers is called an arithmetic progression if the difference between any two consecutive elements is the same.

Given an array of numbers arr, return true if the array can be rearranged to form an arithmetic progression. Otherwise, return false.

 

```python
from typing import List
```

```python
class Solution:
    def canMakeArithmeticProgression(self, arr: List[int]) -> bool:
        arr.sort()
        check = arr[1]-arr[0]
        for i in range(len(arr)-1,1,-1):
            if(check != arr[i]-arr[i-1]):
                return  False
        
        return True
```

```python
solution = Solution()
result = solution.canMakeArithmeticProgression([4,5.09,6])
print(result)
```

Time complexity O(n  logn)
Space Complexity
 uses O(1) extra space, but Python's built-in Timsort may require up to O(n) auxiliary memory internally.
