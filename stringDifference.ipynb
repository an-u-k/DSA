You are given two strings s and t.

String t is generated by random shuffling string s and then add one more letter at a random position.

Return the letter that was added to t.

```python
class Solution:
    def findTheDifference(self, s: str, t: str) -> str:
        a=''
        for x,y in zip(sorted(s),sorted(t)):
            if(x!=y):
                return y
            
        return sorted(t)[-1]

        
        
```

```python
solution = Solution()
result = solution.findTheDifference("hello","heillo")
print(result)
```

Time Complexity
sorted(s) → O(n log n)
sorted(t) inside zip → O(n log n)
zip(...) iteration → O(n) in the worst case.
sorted(t)[-1] at the end sorts t again → O(n log n).

So the total is:
O(nlogn)+O(nlogn)+O(n)+O(nlogn)=O(nlogn)


Space Complexity
sorted(s) creates a new list of size n → O(n)
sorted(t) creates a new list of size n+1 → O(n)

So the auxiliary space is O(n)
